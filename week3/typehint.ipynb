### Typehint 연습
이 파일은 파이썬의 typehint 예제와 이론을 실습한 결과입니다. 작성자: 권순욱 작성일: 2026-09-15

타입힌트는 함수의 매개변수와 반환값의 타입을 지정하는 목적으로 사용하며, 엄격한 타입지정으로 인한 프로그래머의 의도를 명확히 전달하기 위함이다. 안써도 되지만 파이썬 13부터는 쓰는 것을 권장하고 있다.

```python
# 함수 타입힌트 연습
# greet 함수 실습


# 1. 기본적인 함수 만들기
def greet(msg):
    print("in greet:", msg)
    return msg + "!"


print(greet("안녕"))


# 2. 함수에 문자열을 넣어보기
print(greet("안녕하세요"))
print(greet("반가워"))
print(greet("파이썬 공부 중입니다"))


# 3. 타입힌트를 사용한 greet 함수
def greet(msg: str) -> str:
    print("in greet:", msg)
    return msg + "!"


print(greet("안녕"))


# 4. 여러 문자열을 넣어서 테스트
print(greet("좋은 아침"))
print(greet("반갑습니다"))
print(greet("오늘도 열심히 공부하자"))
print(greet("Python"))


# 5. 변수에 문자열을 저장해서 함수에 전달
message1: str = "영준"
message2: str = "정아"
message3: str = "민재"
message4: str = "완"

print(greet(message1))
print(greet(message2))
print(greet(message3))
print(greet(message4))


# 6. 반환값을 변수에 저장
result: str = greet("안녕하세요")

print("반환된 값:", result)


# 7. 함수의 반환값을 다른 문자열과 사용
result1 = greet("Hello")
result2 = greet("Python")

print(result1)
print(result2)


# 8. 문자열을 여러 번 테스트
print(greet("하나"))
print(greet("둘"))
print(greet("셋"))
print(greet("넷"))
print(greet("다섯"))


# 9. 타입힌트 확인
# msg: str
# -> msg는 문자열을 받는다고 표시
#
# -> str
# -> 함수가 문자열을 반환한다고 표시


# 10. 타입힌트가 있어도 실행할 때 자동으로 타입을 검사하지는 않는다.
# 따라서 아래처럼 숫자를 넣는 것 자체는 가능하다.
#
# greet(3.14)
#
# 하지만 함수 안에서
# msg + "!"
# 를 실행할 때 float와 str을 더할 수 없기 때문에
# TypeError가 발생한다.


# 11. 숫자를 넣었을 때 발생하는 오류 실습
# 아래 주석을 풀면 TypeError가 발생한다.

# print(greet(3.14))


# 12. 정수도 같은 이유로 오류가 발생한다.
# print(greet(10))


# 13. 문자열로 변환하면 정상적으로 실행된다.
number: int = 10

print(greet(str(number)))


# 14. 숫자를 문자열로 변환해서 함수에 전달
age: int = 20

age_message: str = greet(str(age))

print(age_message)


# 15. 최종적으로 타입힌트를 사용한 함수
def greet(msg: str) -> str:
    print("in greet:", msg)
    return msg + "!"


# 함수 호출
print(greet("안녕"))
print(greet("반가워"))
print(greet("파이썬"))


# 16. 반환값 확인
message: str = greet("공부")

print("최종 결과:", message)


# 17. 여러 학생 이름으로 테스트
students: list[str] = ["영준", "정아", "민재", "완"]

for student in students:
    print(greet(student))
```
