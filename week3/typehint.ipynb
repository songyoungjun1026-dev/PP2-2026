이 파일은 파이썬의 typehint 예재와 이론을 실습한 결과입니다.

타입힌트는 함수의 매개변수와 반환값에 타입을 지정하는 목적으로 사용한며, 엄격한 타입지정으로 인한 프로그래머 의도를 명확하게 전달하기 위함이다, 안써도 되지만 파이썬 13부터는 쓰는 것을 권장하고 있다.

```python
def greet(msg : str) -> None:
    print("in greet:", msg)
    return msg + "!"

print(greet("안녕"))
greet(3.141592)

# 타입 힌트 연습

name: str = "철수"
age: int = 20
height: float = 175.5
student: bool = True

print(name)
print(age)
print(height)
print(student)

a: int = 10
b: int = 20

print(a + b)

x: float = 3.14
y: float = 2.5

print(x + y)

first_name: str = "김"
last_name: str = "철수"

print(first_name + last_name)

numbers: list[int] = [1, 2, 3, 4, 5]
names: list[str] = ["철수", "영희", "민수"]

print(numbers)
print(names)

def add(a: int, b: int) -> int:
    return a + b

print(add(10, 20))

def hello(name: str) -> str:
    return "안녕하세요 " + name

print(hello("철수"))

def multiply(a: float, b: float) -> float:
    return a * b

print(multiply(2.5, 3.0))

def greeting(name: str) -> None:
    print("안녕하세요", name)

greeting("영희")

def total(numbers: list[int]) -> int:
    return sum(numbers)

print(total([1, 2, 3, 4, 5]))

age: int = 21
price: float = 1500.5
title: str = "Python"
is_student: bool = False

print(age)
print(price)
print(title)
print(is_student)

person: dict[str, str] = {
    "name": "철수",
    "city": "서울"
}

print(person)

def print_names(names: list[str]) -> None:
    for name in names:
        print(name)

print_names(["철수", "영희", "민수"])
```
