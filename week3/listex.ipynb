List 연습 노트 파일입니다.

리스트의 합병 연산 (+)
리스트의 곱 연산(*)

```python
# 리스트의 합병 연산 연습

a = [1, 2, 3]
b = [4, 5, 6]

print(a + b)

x = ["사과", "바나나"]
y = ["포도", "딸기"]

print(x + y)

print([1, 2] + [3, 4])
print([10, 20] + [30, 40])
print(["a", "b"] + ["c", "d"])


a = [1, 2, 3]

print(a * 2)
print(a * 3)

b = ["안녕"]
print(b * 4)

print([0] * 5)
print([1, 2] * 3)


a = [1, 2]
b = [3, 4]

print(a + b)
print(a * 2)
print(b * 3)

print((a + b) * 2)
print((a * 2) + b)


list1 = [10, 20]
list2 = [30, 40]

print(list1 + list2)
print(list1 * 3)
print(list2 * 2)

list3 = ["python", "java"]
list4 = ["C", "C++"]

print(list3 + list4)
print(list3 * 2)


empty = []
print(empty + [1, 2, 3])
print(empty * 5)


print(["a", "b"] * 3)
print(["hello"] * 2)


num = [1, 2, 3]
print(num + [4, 5])
print(num * 2)

num2 = [5, 10]
print(num2 + [15, 20])
print(num2 * 4)
```
