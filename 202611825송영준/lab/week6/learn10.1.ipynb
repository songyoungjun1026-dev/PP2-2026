1. 학습 목표 및 개요

* 파일 입출력: 텍스트 파일과 이진 파일의 개념을 이해하고 읽고 쓰기를 수행할 수 있다.
* 디렉토리와 경로: 파일 시스템의 절대/상대 경로 및 디렉토리 탐색(os 모듈)을 다룰 수 있다.
* 정규식(Regular Expressions): re 모듈을 이용해 특정 규칙을 가진 문자열을 검색하고 추출할 수 있다.
* CSV 및 피클(Pickle): 구조화된 데이터(CSV)와 파이썬 객체 저장/복원(Pickle)을 활용할 수 있다.
* 예외 처리(Exception Handling): 프로그램 오류 발생 시 try-except-finally를 통해 안전하게 처리할 수 있다.

2. 핵심 이론 및 개념 상세 노트

2-1. 파일과 디렉토리 개념
 
 * 파일 포인터 (Offset): 파일을 열면 포인터가 0에서 시작하며, 읽고 쓸 때마다 접근한 바이트 수만큼 이동합니다. 파일의 끝은 **EOF (End of File)**라고 합니다.
 * 텍스트 파일: 문자로 구성된 파일 (UTF-8 등 인코딩/디코딩 과정 거침) , 이진 파일: 이진수 바이트 배열이 그대로 저장된 파일 ("rb", "wb" 모드 사용)
 * 절대 경로: 최상위 루트부터 시작하는 전체 경로 (예: C:\user\song\...), 상대 경로: 현재 작업 중인 폴더를 기준(. 또는 ..)으로 하는 경로

2-2. 예외 처리 (try-except)

 * 프로그램 실행 중 오류(ZeroDivisionError, ValueError, IOError 등)가 발생했을 때 비정상 종료(Crash)를 막고 우아하게 처리합니다.
 * finally 블록은 예외 발생 여부와 상관없이 무조건 실행되므로 파일 닫기(close()) 등에 주로 사용됩니다.
