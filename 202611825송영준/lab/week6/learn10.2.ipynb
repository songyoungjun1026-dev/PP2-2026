```python
infilename = input("입력 파일 이름: ")  # sales.txt 입력
outfilename = input("출력 파일 이름: ")  # summary.txt 입력
try:
    infile = open(infilename, "r")  # try catch 로 묶어야 함! 없을 경우 에러
    outfile = open(outfilename, "w")
    sum = 0
    count = 0

    line = infile.readline()

    while line != "":
        s = int(line.strip())  # 문자열을 정수형으로 바꾸기
        sum += s
        count += 1
        line = infile.readline()

    outfile.write("총매출 = " + str(sum) + "\n")
    outfile.write("평균 일매출 = " + str(sum / count))

    infile.close()
    outfile.close()
    
except FileNotFoundError:
    print("파일 찾을 수 없음")
```
