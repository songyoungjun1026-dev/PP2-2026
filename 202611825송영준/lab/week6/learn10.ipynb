# 작업폴더 : python 파일 실행하는 위치


```python
with open("file.txt", "r", encoding = "utf-8") as f:
    for line in f:
        print(line.strip())

    line = f.readline().strip()
    while line != "":
        print(line)
        line = f.readline().strip()

    eline = f.readline().strip()
    print(eline)
    print(f.tell())
```

# 파일 읽기와 쓰기 (기본 및 with문)

```python
# 1. 수동 파일 열고 닫기
infile = open("input.txt", "r", encoding="utf-8")
line = infile.readline()
while line != "":
    print(line.rstrip())
    line = infile.readline()
infile.close()

# 2. with 문 사용 (자동으로 파일 닫기 수행)
with open("test.txt", "w", encoding="utf-8") as f:
    f.write("김영희\n")
    f.write("최자영\n")
```

# Lab: 매출 파일 처리

```python
# sales.txt를 읽어 총매출과 평균 일 매출을 계산해 summary.txt에 저장
infilename = input("입력 파일 이름: ")  # 예: sales.txt
outfilename = input("출력 파일 이름: ")  # 예: summary.txt

try:
    infile = open(infilename, "r", encoding="utf-8")
    outfile = open(outfilename, "w", encoding="utf-8")

    total_sum = 0
    count = 0

    for line in infile:
        s = int(line.strip())
        total_sum += s
        count += 1

    outfile.write("총매출= " + str(total_sum) + "\n")
    outfile.write("평균일매출= " + str(total_sum / count) + "\n")

    infile.close()
    outfile.close()
    print("매출 파일 처리가 완료되었습니다.")

except IOError:
    print("파일을 찾을 수 없습니다.")
```

# CSV 파일 처리 (날씨 데이터 최저기온 찾기)

```python
import csv

# weather.csv 파일에서 최저기온(3번 인덱스)이 가장 낮은 값 찾기
try:
    f = open('weather.csv', 'r', encoding='utf-8')
    data = csv.reader(f)
    header = next(data)  # 헤더 건너뛰기
    
    min_temp = 1000
    for row in data:
        current_temp = float(row[3])
        if min_temp > current_temp:
            min_temp = current_temp
            
    print(f'가장 추웠던 날의 기온은 {min_temp}℃ 입니다.')
    f.close()
except IOError:
    print("weather.csv 파일을 열 수 없습니다.")
```

# 디렉토리 및 파일 검색 (os 모듈)

```python
import os

# 현재 작업 디렉토리의 파일 중 확장자가 '.jpg'인 파일 찾기
cwd = os.getcwd()
print(f"현재 작업 디렉토리: {cwd}")

files = os.listdir()
for name in files:
    if os.path.isfile(name):
        if name.endswith(".jpg"):
            print("발견된 이미지:", name)
```

이진 파일 복사 (Binary Copy) 

```python
# 이미지 등 이진 파일을 1024바이트씩 읽어서 복사하기
infile = open("123.png", "rb")
outfile = open("kkk.png", "wb")

while True:
    copy_buffer = infile.read(1024)
    if not copy_buffer:
        break
    outfile.write(copy_buffer)

infile.close()
outfile.close()
print("파일 복사가 완료되었습니다.")
```

# 객체 직렬화/역직렬화 (pickle 모듈)

```python
import pickle

# 1. 딕셔너리 객체를 이진 파일로 저장 (Dump)
gameOption = {
    "Sound": 8,
    "VideoQuality": "HIGH",
    "Money": 100000,
    "WeaponList": ["gun", "missile", "knife"]
}

with open("save.p", "wb") as file:
    pickle.dump(gameOption, file)

# 2. 이진 파일에서 객체 복원하기 (Load)
with open("save.p", "rb") as file:
    obj = pickle.load(file)
    print("복원된 객체:", obj)
```

# 정규식(re) 활용 – 패스워드 검사기

```python
import re

password = input("패스워드를 입력하세요: ")
flag = 0

while True:
    if len(password) < 8:
        flag = -1
        break
    elif not re.search("[a-z]", password):
        flag = -1
        break
    elif not re.search("[A-Z]", password):
        flag = -1
        break
    elif not re.search("[0-9]", password):
        flag = -1
        break
    elif not re.search("[_@$]", password):
        flag = -1
        break
    else:
        flag = 0
        break

if flag == 0:
    print("유효한 패스워드입니다!")
else:
    print("유효한 패스워드가 아닙니다. (조건: 8글자 이상, 대/소문자, 숫자, 특수문자[_, @, $] 포함)")
```

예외 처리 기본 구조 (try-except-finally)

```python
# 0으로 나누기 예외 처리 예제
x, y = 2, 0

try:
    z = x / y
except ZeroDivisionError as e:
    print("예외 발생:", e)
finally:
    print("예외 처리 블록이 종료되었습니다.")
```
