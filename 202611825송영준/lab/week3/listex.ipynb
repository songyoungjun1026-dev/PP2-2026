### LIST 연습 노트 파일입니다.
리스트의 합병 연산 (+) 리스트의 곱 연산 (*)

```python
# LIST 연습 노트
# 리스트의 합병 연산 (+), 리스트의 곱 연산 (*)

# 리스트 합병 연산
list1 = ["영준", "정아"]
list2 = ["민재", "완"]

result1 = list1 + list2

print("list1:", list1)
print("list2:", list2)
print("합병 결과:", result1)


# 리스트 곱 연산
list3 = [1, 2, 3]

result2 = list3 * 2

print("list3:", list3)
print("곱 연산 결과:", result2)


# 문자열 리스트도 반복 가능
names = ["영준", "정아"]

print("이름 반복:", names * 3)
```
