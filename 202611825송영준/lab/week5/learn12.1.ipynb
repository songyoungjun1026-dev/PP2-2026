1. 학습 목표 (Learning Objectives)

 * 부모 클래스를 상속받아서 새로운 자식 클래스를 정의할 수 있다.
 * 부모 클래스의 메서드를 자식 클래스에서 재정의(메소드 오버라이딩)할 수 있다.
 * Object 클래스의 개념과 역할을 이해할 수 있다.
 * 메서드 오버라이딩을 활용하여 다형성을 이해하고 사용할 수 있다.
 * 클래스 간의 관계(is-a 관계 vs has-a 관계)를 파악하고 설계할 수 있다.

 2. 이번 장에서 다루는 핵심 내용 요약
 
 * 상속(Inheritance)과 is-a 관계
 * 기존 클래스(부모)의 코드와 데이터를 자식 클래스가 물려받아 확장하는 기법입니다.
 * 객체지향에서 상속은 "is-a" 관계를 나타냅니다. (예: 푸들은 강아지이다, 승용차는 차량이다.) 
 * 자식 클래스에서 super().__init__(...)을 통해 반드시 부모 생성자를 호출해야 속성들이 정상적으로 초기화됩니다.
 
2-1. 접근 제어와 가시성 (Visibility)

* 파이썬은 엄격한 제어자는 없지만 관례적으로 밑줄 기호를 사용합니다.
* public (밑줄 없음): 어디서나 접근 가능
* protected (_value): 하위 클래스 접근 권장, 외부 직접 접근 지_
* __private (__value): 이름 맹글링(name mangling)을 통해 클래스 내부에서만 접근 가능 (자식 클래스 접근 불가)

2-2. 메소드 오버라이딩 (Overriding)과 다형성

* 자식 클래스가 부모 클래스의 메서드를 동일한 이름으로 재정의하여 기능을 변경/확장합니다.
* 다형성(Polymorphism): 같은 이름의 메서드를 호출하더라도 객체의 클래스에 따라 다른 동작을 수행하는 객체지향의 핵심 성질입니다.

2-3. 추상 클래스 (Abstract Class)와 추상 메서드
* abc 모듈의 ABC와 @abstractmethod 데코레이터를 사용하여 구현합니다.
* 추상 메서드는 하위 클래스에서 반드시 구현해야 하며, 추상 클래스는 직접 인스턴스(객체)를 생성할 수 없습니다.

2-4. Dunder 메서드 (특수 메서드)
* Double Underlines (__xxx__)의 약자로, 파이썬 내부에서 자동 호출되는 특별한 메서드입니다.
* 예: __init__(생성자), __str__(문자열 표현), __eq__(동등 비교), __del__(소멸자) 등

2-5. 클래스 간의 관계
* is-a 관계: 상속 관계 (예: Car is a Vehicle)
* has-a 관계: 구성 관계, 소유 관계 (예: Person has a pet)_
